# simData04 analysis

Notebook analysis for the `simData04` case study.

The code below keeps the workflow explicit so it stays easy to compare with the original R reference.

```python
from pathlib import Path

import matplotlib.pyplot as plt
from glotaran.io import load_dataset, load_model, load_parameters
from glotaran.optimization.optimize import optimize
from glotaran.project.scheme import Scheme
from pyglotaran_extras.plotting.plot_overview import plot_overview
from pyglotaran_extras.plotting.style import PlotStyle

script_dir = Path.cwd()
print(f"Notebook folder: {script_dir}")

plot_style = PlotStyle()
plt.rc("axes", prop_cycle=plot_style.cycler)
plt.rcParams["figure.figsize"] = (21, 14)

dataset = load_dataset(script_dir.joinpath("simData04.ascii"))
dataset
```

## Kinetic model

Custom-axis kinetic fit with IRF and fixed initial concentrations.

```python
print(f"\n{'#' * 10} Kinetic Model {'#' * 10}\n")

kin_model = load_model(script_dir.joinpath("simData4/kin_model_ds4.yml"))
kin_parameters = load_parameters(script_dir.joinpath("simData4/kin_params_ds4.yml"))
kin_model.validate(parameters=kin_parameters)

kin_scheme = Scheme(
    kin_model,
    kin_parameters,
    data={"dataset": dataset},
)
kin_result = optimize(kin_scheme)

print(f"\n{'#' * 3} Kinetic Model - Optimization Result {'#' * 3}\n")
print(kin_result)
print(f"\n{'#' * 3} Kinetic Model - Optimized Parameters {'#' * 3}\n")
print(kin_result.optimized_parameters)

plot_overview(kin_result.data["dataset"], linlog=False)
plt.gcf().suptitle("simData04 kinetic")
plt.show()
```

## Spectral model

Spectral fit using the skewed-Gaussian basis for the two species.

```python
print(f"\n{'#' * 10} Spectral Model {'#' * 10}\n")

spectral_model = load_model(script_dir.joinpath("simData4/spectral_model_ds4.yml"))
spectral_parameters = load_parameters(script_dir.joinpath("simData4/spectral_params_ds4.yml"))
spectral_model.validate(parameters=spectral_parameters)

spectral_scheme = Scheme(
    spectral_model,
    spectral_parameters,
    data={"dataset": dataset},
)
spectral_result = optimize(spectral_scheme)

print(f"\n{'#' * 3} Spectral Model - Optimization Result {'#' * 3}\n")
print(spectral_result)
print(f"\n{'#' * 3} Spectral Model - Optimized Parameters {'#' * 3}\n")
print(spectral_result.optimized_parameters)

plot_overview(spectral_result.data["dataset"], linlog=False)
plt.gcf().suptitle("simData04 spectral")
plt.show()
```

## Spectrotemporal model

Combined kinetic and spectral fit using a generated full-model definition for this case.

```python
print(f"\n{'#' * 10} Spectrotemporal ('Full') Model {'#' * 10}\n")

spectemp_model = load_model(script_dir.joinpath("simData4/full_model_ds4.yml"))
spectemp_parameters = load_parameters(script_dir.joinpath("simData4/full_params_ds4.yml"))
spectemp_model.validate(parameters=spectemp_parameters)

spectemp_scheme = Scheme(
    spectemp_model,
    spectemp_parameters,
    data={"dataset": dataset},
)
spectemp_result = optimize(spectemp_scheme)

print(f"\n{'#' * 3} Spectrotemporal Model - Optimization Result {'#' * 3}\n")
print(spectemp_result)
print(f"\n{'#' * 3} Spectrotemporal Model - Optimized Parameters {'#' * 3}\n")
print(spectemp_result.optimized_parameters)

plot_overview(spectemp_result.data["dataset"], linlog=True, linthresh=80)
plt.gcf().suptitle("simData04 spectrotemporal")
plt.show()
```
